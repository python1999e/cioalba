## 1 · Variables and data types

A **variable** is a labeled box that stores a value. You create one with `=`.
Python has a few basic types you will use constantly:

- **int** — whole number, e.g. `42`
- **float** — decimal number, e.g. `3.14`
- **str** — text ('string'), always in quotes, e.g. `'Xanthomonas'`
- **bool** — `True` or `False`

````python
pathogen = "Xanthomonas oryzae"   # str
num_lesions = 7                    # int
severity_pct = 23.5                # float
is_infected = True                 # bool

print(pathogen, type(pathogen))
print(num_lesions, type(num_lesions))
print(severity_pct, type(severity_pct))
print(is_infected, type(is_infected))
````

### Arithmetic and f-strings
You can do math with numbers, and build readable text with **f-strings** (a string with an
`f` in front, where `{ }` inserts a variable).

````python
leaves_total = 20
leaves_diseased = 7
incidence = leaves_diseased / leaves_total * 100

print(f"Disease incidence: {incidence:.1f}%")   # IMPORTANT f is different from .1f = 1 decimal place
````

## 2 · Lists and dictionaries

Real data comes in collections. The two you need most:

- **list** — an *ordered* sequence, in square brackets `[ ]`.
- **dict** (dictionary) — *key → value* pairs, in curly braces `{ }`.

````python
# A list of disease severity scores (0-9 scale) for 6 plants
scores = [0, 2, 5, 9, 3, 1]

print("All scores:", scores)
print("First score:", scores[0])     # counting starts at 0!
print("Last score:", scores[-1])     # -1 means last
print("How many plants:", len(scores))
print("Average score:", sum(scores) / len(scores))
````

add and extned the list using append

````python
scores.append(7)
print("All scores:", scores)
````

````python
scores.extend([4, 8])
print("All scores:", scores)
````

Notice **indexing starts at 0**: `scores[0]` is the first item. This trips up everyone at
first — it is normal.

````python
# A dictionary: information about one sample
sample = {
    "id": "S-001",
    "host": "rice",
    "pathogen": "Magnaporthe oryzae",
    "severity": 6,
}

print(sample["pathogen"])          # look up a value by its key
sample["severity"] = 8             # change a value
sample["date"] = "2026-08-26"      # add a new key
print(sample)
````

## 3 · Loops and conditions (making decisions)

A **`for` loop** repeats an action for every item in a collection.
An **`if` statement** does something only when a condition is true.

⚠️ VERY IMPORTANT !! Python uses **indentation** (the spaces at the start of a line) to know what is *inside*
the loop or `if`. Colab adds it automatically after a `:`.

````python
scores = [0, 2, 5, 9, 3, 1]

for s in scores:
    if s >= 5:
        print(f"Score {s}: SEVERE — needs attention")
    elif s >= 1:
        print(f"Score {s}: mild")
    else:
        print(f"Score {s}: healthy")
````

### Building a new list with a loop
A very common pattern: go through data and collect results.

````python
scores = [0, 2, 5, 9, 3, 1]

severe_plants = []          # start with an empty list
for s in scores:
    if s >= 5:
        severe_plants.append(s)   # add to the list

print("Severe scores:", severe_plants)
print("Number of severe plants:", len(severe_plants))
````

## 4 · Functions (reusable recipes)

A **function** packages steps so you can reuse them. Define with `def`, send in **inputs**
(arguments), and get back a result with `return`.

````python
def disease_incidence(diseased, total):
    """Return the % of plants that are diseased."""
    return diseased / total * 100

# Now use it as many times as you like
print(disease_incidence(7, 20))
print(disease_incidence(15, 60))

result = disease_incidence(3, 12)
print(f"Incidence = {result:.1f}%")
````

## 6 · pandas — working with tables of data

**pandas** is the workhorse of data analysis. Its main object is the **DataFrame**: a table
with rows and named columns, just like a spreadsheet. This is where the course really lives.

Let's build a small dataset of plant-disease field observations.

````python
data = {
    "sample_id": ["S1", "S2", "S3", "S4", "S5", "S6", "S7", "S8"],
    "host":      ["rice", "rice", "maize", "maize", "tomato", "tomato", "rice", "maize"],
    "pathogen":  ["M. oryzae", "X. oryzae", "F. verticillioides", "Puccinia",
                  "P. infestans", "Alternaria", "M. oryzae", "F. verticillioides"],
    "severity":  [6, 8, 3, 5, 9, 2, 7, 4],          # 0-9 scale
    "temp_C":    [26, 28, 24, 22, 18, 30, 27, 25],
    "treated":   [False, True, False, True, True, False, False, True],
}

df = pd.DataFrame(data)
df   # in Colab, the last line of a cell is displayed automatically
````

### First things to always do with a new dataset
Get a feel for its size, columns, and summary statistics.

````python
print("Shape (rows, columns):", df.shape)
print()
print("Column names:", list(df.columns))
print()
df.info()      # data type of each column + missing values
````

````python
df.describe()   # summary stats for the numeric columns
````

### Selecting columns and rows

````python
# One column (a Series)
print(df["severity"])
print()
# Several columns (a smaller DataFrame)
df[["host", "severity"]]
````

````python
# The first few rows
df.head(3)
````

### Filtering — the most useful skill
Keep only the rows that meet a condition. Read `df[df["severity"] >= 6]` as
*"the rows of df where severity is at least 6"*.

````python
# Severe cases
severe = df[df["severity"] >= 6]
severe
````

````python
# Combine conditions with & (and) / | (or) — each condition in ( )
rice_severe = df[(df["host"] == "rice") & (df["severity"] >= 6)]
rice_severe
````

### Creating a new column
Columns are computed all at once (vectorized) — no loop.

````python
# Flag high-severity samples
df["high_severity"] = df["severity"] >= 6

# Convert temperature to Fahrenheit
df["temp_F"] = df["temp_C"] * 9/5 + 32
df
````

### Grouping and summarizing
`groupby` answers questions like *"what is the average severity per host?"* — this is the
heart of most data analysis.

````python
# Average severity for each host
df.groupby("host")["severity"].mean()
````

````python
# Several statistics at once, per host
df.groupby("host")["severity"].agg(["count", "mean", "max"])
````

````python
# How many samples of each host?
df["host"].value_counts()
````

## 8 · Making a plot

A picture beats a table. pandas can plot directly, using matplotlib underneath.

````python
# Bar chart: average severity per host
avg_sev = df.groupby("host")["severity"].mean()

avg_sev.plot(kind="bar", color="seagreen")
plt.title("Average disease severity by host")
plt.ylabel("Severity (0-9)")
plt.xlabel("Host")
plt.xticks(rotation=0)
plt.show()
````

````python
# Scatter plot: does temperature relate to severity?
plt.scatter(df["temp_C"], df["severity"], color="darkorange")
plt.title("Severity vs temperature")
plt.xlabel("Temperature (°C)")
plt.ylabel("Severity (0-9)")
plt.show()
````
